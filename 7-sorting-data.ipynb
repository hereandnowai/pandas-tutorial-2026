# Part 7: Sorting Data

In this section, we'll learn how to sort our DataFrames by column values. We'll start with a simple dictionary-based example and then move to our real-world survey dataset.

```python
people = {
    "first": ["Rohit", "Virat", "Jasprit", "Shubman", "Hardik", "Rishabh", "Suryakumar", "Ravindra", "Kuldeep", "Arshdeep", "Axar"],
    "last": ["Sharma", "Kohli", "Bumrah", "Gill", "Pandya", "Pant", "Yadav", "Jadeja", "Yadav", "Singh", "Patel"],
    "email": [
        "rohit.sharma@email.com", 
        "virat.kohli@email.com", 
        "jasprit.bumrah@email.com", 
        "shubman.gill@email.com", 
        "hardik.pandya@email.com", 
        "rishabh.pant@email.com", 
        "suryakumar.yadav@email.com", 
        "ravindra.jadeja@email.com", 
        "kuldeep.yadav@email.com", 
        "arshdeep.singh@email.com", 
        "axar.patel@email.com"
    ]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.sort_values(by='last')
```

## Sorting by a Single Column
Use the `sort_values()` method to sort by a column. By default, it sorts in ascending order.

```python
df.sort_values(by='last', ascending=False)
```

To sort in descending order, set `ascending=False`.

```python
df.sort_values(by=['last', 'first'], ascending=[True, False])
```

## Sorting by Multiple Columns
We can sort by multiple columns by passing a list. Let's sort by 'last' name ascending and then 'first' name descending.

```python
df.sort_index()
```

## Sorting by Index
If you want to return the DataFrame to its original state or sort by labels, use `sort_index()`.

```python
df['last'].sort_values()
```
