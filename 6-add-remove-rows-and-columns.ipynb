# Part 6: Adding and Removing Rows/Columns

In this section, we'll learn how to modify the structure of our DataFrame by creating new columns, dropping existing ones, and adding/removing rows.

```python
people = {
    "first": ["Rohit", "Virat", "Jasprit", "Shubman", "Hardik", "Rishabh", "Suryakumar", "Ravindra", "Kuldeep", "Arshdeep", "Axar"],
    "last": ["Sharma", "Kohli", "Bumrah", "Gill", "Pandya", "Pant", "Yadav", "Jadeja", "Yadav", "Singh", "Patel"],
    "email": [
        "rohit.sharma@email.com", 
        "virat.kohli@email.com", 
        "jasprit.bumrah@email.com", 
        "shubman.gill@email.com", 
        "hardik.pandya@email.com", 
        "rishabh.pant@email.com", 
        "suryakumar.yadav@email.com", 
        "ravindra.jadeja@email.com", 
        "kuldeep.yadav@email.com", 
        "arshdeep.singh@email.com", 
        "axar.patel@email.com"
    ]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['first'] + ' ' + df['last']
```

## Creating a New Column
We can create new columns by performing operations on existing ones. Here, we're combining 'first' and 'last' name into a single 'full_name' column.

```python
df['full_name'] = df['first'] + ' ' + df['last']
```

```python
df
```

```python
df.drop(columns=['first', 'last'], inplace=True)
```

## Removing Columns
To remove columns, use the `drop()` method with the `columns` argument. Setting `inplace=True` makes the change permanent in the current DataFrame.

```python
df
```

```python
df['full_name'].str.split(' ', expand=True)
```

## Splitting Columns
We can split a single column into multiple columns using `.str.split()`. Using `expand=True` returns a DataFrame instead of a list.

```python
df[['first', 'last']] = df['full_name'].str.split(' ', expand=True)
```

```python
df
```

```python
new_row = pd.DataFrame([{'first': 'Abisheik'}])
df = pd.concat([df, new_row], ignore_index=True)
```

## Adding Rows
To add rows, we can use `pd.concat()` to join two DataFrames together. `ignore_index=True` ensures the resulting DataFrame has a clean, continuous index.

```python
df
```

```python
df = pd.concat([df, new_row], ignore_index=True)
```

```python
df
```

```python
df.drop(index=12, inplace=True)
```

## Removing Rows
Rows can be removed by their index using the `drop()` method.

```python
df
```

```python
remove_yadav = df['last'] == 'Yadav'
df.drop(index=df[remove_yadav].index, inplace=True)
```

You can also drop rows based on a condition by finding the indices that match the condition.

```python
df
```
