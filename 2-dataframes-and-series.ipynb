## Accessing Columns
Columns are actually **Series** objects. You can access them using bracket notation (recommended) or dot notation.

```python
type(df['email'])
```

```python
df.email
```

```python
df[['last', 'email']]
```

To select multiple columns, pass a **list** of column names inside the brackets. This returns a new DataFrame.

```python
type(df[['last', 'email']])
```

```python
df.columns
```

```python
df.iloc[0]
```

## Selecting Rows: iloc vs. loc
Pandas provides two main ways to select rows:
1. `iloc`: Integer-location based indexing (using 0-based positions).
2. `loc`: Label-based indexing (using names/labels).

```python
type(df.iloc[0])
```

```python
df.iloc[[0,1]]
```

```python
df.iloc[[0,1], 2]
```

```python
df.loc[0]
```

```python
df.loc[[0,1], 'email']
```

```python
df.loc[[0,1], ['email', 'last']]
```
