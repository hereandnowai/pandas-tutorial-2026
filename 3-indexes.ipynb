# Part 3: Indexes - Setting, Resetting, and Using Custom Indexes

In this section, we'll learn how to change the index of our DataFrame. By default, pandas uses a RangeIndex (0, 1, 2...), but often it's more useful to use a specific column (like an email or a unique ID) as the index for easier lookups.

```python
import pandas as pd
```

```python
people = {
    "first": ["Rohit", "Virat", "Jasprit", "Shubman", "Hardik", "Rishabh", "Suryakumar", "Ravindra", "Kuldeep", "Arshdeep", "Axar"],
    "last": ["Sharma", "Kohli", "Bumrah", "Gill", "Pandya", "Pant", "Yadav", "Jadeja", "Yadav", "Singh", "Patel"],
    "email": [
        "rohit.sharma@email.com", 
        "virat.kohli@email.com", 
        "jasprit.bumrah@email.com", 
        "shubman.gill@email.com", 
        "hardik.pandya@email.com", 
        "rishabh.pant@email.com", 
        "suryakumar.yadav@email.com", 
        "ravindra.jadeja@email.com", 
        "kuldeep.yadav@email.com", 
        "arshdeep.singh@email.com", 
        "axar.patel@email.com"
    ]
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['email']
```

```python
df.set_index('email')
```

## Setting a Custom Index
We can use `set_index()` to move a column into the index position. By default, this returns a *new* DataFrame and doesn't modify the original one.

```python
df
```

```python
df.set_index('email', inplace=True)
```

To modify the DataFrame permanently, we use `inplace=True`. Now, lookups can be done using the email addresses instead of row numbers.

```python
df
```

```python
df.index
```

```python
df.loc['rohit.sharma@email.com']
```

## Index-Based Selection
Now that our index is set to the 'email' column, we can use `loc` to fetch rows using specific email addresses.

```python
df.loc['rohit.sharma@email.com', 'last']
```

```python
df.loc[0]
```

```python
df.iloc[0]
```

```python
df.reset_index(inplace=True)
```

## Fixing the KeyError
You encountered a `KeyError: 'AIHuman'` because the `schema_df` is currently using row numbers (0, 1, 2...) as the index, so it doesn't recognize `'AIHuman'` as a valid row label.

To fix this, we need to set the index of `schema_df` to the column that contains the attribute names (usually 'qname').

```python
# First, set the index to 'qname' so we can look up rows by the question name
schema_df.set_index('qname', inplace=True)

# Now we can look up 'AIHuman' without a KeyError!
schema_df.loc['AIHuman']
```

## Understanding the KeyError: 'AIOpen'
You might encounter another `KeyError` if you try to access a question like `'AIOpen'` directly without specifying `loc`. 

In pandas, if you use `df['Label']`, it looks for a **Column**. Since `'AIOpen'` is now a value in our **Index** (the row labels), we must use `.loc[]` to tell pandas we are looking for a row.

```python
# Use .loc to look up the row label 'AIOpen' in the index
schema_df.loc['AIOpen']
```

```python
schema_df.loc['AIHuman', 'AIOpen']
```
